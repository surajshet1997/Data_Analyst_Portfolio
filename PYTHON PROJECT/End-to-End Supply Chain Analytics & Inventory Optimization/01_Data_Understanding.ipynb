```python
import pandas as pd
import numpy as np

# File path
file_path = "supply_chain_dataset.xlsx"

# Load Excel workbook
excel_file = pd.ExcelFile(file_path)

# Display sheet names
excel_file.sheet_names
```

```python
# Load the actual supply chain dataset

df = pd.read_excel(
    file_path,
    sheet_name="supply_chain_dataset1"
)

# Display first 5 rows
df.head()
```

```python
df.columns.tolist()
```

```python
# Check dataset structure and data types
df.info()
```

```python
# Check dataset dimensions
print(f"Number of rows    : {df.shape[0]:,}")
print(f"Number of columns : {df.shape[1]}")
```

```python
# Check for missing values

missing_values = df.isnull().sum()

print("Missing values by column:")
print(missing_values)

# Total missing values in the dataset

print(f"\nTotal missing values: {df.isnull().sum().sum():,}")
```

```python
# Check for duplicate records

duplicate_count = df.duplicated().sum()

print(f"Number of duplicate records: {duplicate_count:,}")
```

```python
# Check unique values for key categorical columns

categorical_columns = [
    "SKU_ID",
    "Warehouse_ID",
    "Supplier_ID",
    "Region"
]

for column in categorical_columns:
    print(f"{column}: {df[column].nunique():,} unique values")
```

```python
# Display the actual unique values

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
# Identify numerical columns

numeric_columns = df.select_dtypes(include="number").columns.tolist()

print("Numerical columns:")
print(numeric_columns)
```

```python
# Check for negative values in numerical columns

negative_values = (df[numeric_columns] < 0).sum()

print("Negative values by column:")
print(negative_values)
```

```python
# Check unique values in binary flag columns

flag_columns = [
    "Promotion_Flag",
    "Stockout_Flag"
]

for column in flag_columns:
    print(f"{column}: {sorted(df[column].unique())}")
```

```python
# Check date consistency

min_date = df["Date"].min()
max_date = df["Date"].max()
unique_dates = df["Date"].nunique()

print(f"Earliest date : {min_date.date()}")
print(f"Latest date   : {max_date.date()}")
print(f"Unique dates  : {unique_dates:,}")
```

```python
# Check for missing dates in the date range

expected_dates = pd.date_range(
    start=df["Date"].min(),
    end=df["Date"].max(),
    freq="D"
)

missing_dates = expected_dates.difference(df["Date"].unique())

print(f"Expected number of dates : {len(expected_dates):,}")
print(f"Actual number of dates   : {df['Date'].nunique():,}")
print(f"Missing dates            : {len(missing_dates):,}")

if len(missing_dates) > 0:
    print("\nMissing dates:")
    print(missing_dates)
```

```python
# Create a data quality summary

data_quality_summary = pd.DataFrame({
    "Check": [
        "Missing Values",
        "Duplicate Records",
        "Negative Values",
        "Promotion Flag Validation",
        "Stockout Flag Validation",
        "Missing Dates"
    ],
    "Result": [
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        negative_values.sum(),
        set(df["Promotion_Flag"].unique()).issubset({0, 1}),
        set(df["Stockout_Flag"].unique()).issubset({0, 1}),
        len(missing_dates)
    ]
})

data_quality_summary
```
